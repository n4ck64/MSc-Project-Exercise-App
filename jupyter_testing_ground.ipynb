```python
from ollama import generate
from ollama import chat
```

```python
response = generate("openmed-model:latest", "I feel pain in my shoulder when i bench press. Why is that? Also, what would be some good exercises to do while that passes?")
print(response['response'])
```

```python
response = generate("medical-expert:latest", "what is the latissimus dorsi also known as in everyday speech, is it the lats?")
print(response['response'])
```

```python

system_prompt = """You are a medical expert that provides advise on physiotherapy and whether exercises are safe or not
for people with a past history of injury. You do not shy away from answering questions. Do not provide an introduction"""

final_response = chat("openmed-model:latest",
                      messages=[
                          {"role": "system", "content": system_prompt},
                          {"role": "user", "content": "I feel pain in my shoulder when i bench press. Why is that? Also, what would be some good exercises to do while that passes?"}
                      ],
                      options={
                          "temperature": 0.1,
                          "num_predict": 1024,
                          "num_ctx": 4096
                      },
                      stream=True)

print("\nMy response: ", end="")
for chunk in final_response:
    content = chunk.message.content
    print(content, end='', flush=True)
print()
```
